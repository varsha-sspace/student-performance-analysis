```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
data = {
    "Name": ["varsha","khushi","kshama","tanisha","pari"],
    "Python":[85,90,70,87,88],
    "Maths":[78,88,90,76,87],
    "Eco": [89,87,78,66,76]
}
df = pd.DataFrame(data)
print(df)
```

```python
df["Average"] = df[["Python","Maths","Eco"]].mean(axis=1)
print(df)
```

```python
print("Highest Average:",df["Average"].max())
print("Lowest Average:",df["Average"].min())
print("Class Average:",df["Average"].mean())
```

```python
topper = df.loc[df["Average"].idmax()]
print("Class Topper:",topper["Name"])
print("Topper's Average:", topper["Average"])
```

```python
print("Python Average:", df["Python"].mean())

print("Maths Average:", df["Maths"].mean())

print("Economics Average:", df["Eco"].mean())
```

```python
subjects = ["Python", "Maths", "Eco"]

averages = [
    df["Python"].mean(),
    df["Maths"].mean(),
    df["Eco"].mean()
]

plt.bar(subjects, averages)

plt.title("Subject-wise Average Performance")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")

plt.show()
```

```python
df["Result"] = df[["Python", "Maths", "Eco"]].ge(40).all(axis=1)

df["Result"] = df["Result"].map({
    True: "Pass",
    False: "Fail"
})

print(df[["Name", "Result"]])
```

```python
total_students = len(df)

passed_students = (df["Result"] == "Pass").sum()

pass_percentage = (passed_students / total_students) * 100

print("Total Students:", total_students)
print("Passed Students:", passed_students)
print("Pass Percentage:", pass_percentage, "%")
```

```python
result_counts = df["Result"].value_counts()

plt.figure(figsize=(6, 6))

plt.pie(
    result_counts,
    labels=result_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Student Pass and Fail Percentage")

plt.show()
```

**Project Conclusion**

This project analysed the academic performance of five students using Python, NumPy, Pandas and Matplotlib. We calculated individual averages, subject-wise averages, identified the topper and analysed pass percentages.

The analysis showed that all five students passed, resulting in a 100% pass rate
